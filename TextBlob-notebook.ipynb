## Creating a TextBlob

First, I imported TextBlob and created a simple TextBlob object.

```python
from textblob import TextBlob
```

```python
wiki = TextBlob("Python is a high-level, general-purpose programming language.")
wiki
```

## Part-of-Speech Tagging

The `.tags` property labels each word based on its part of speech, like noun, adjective, or verb.

```python
wiki.tags
```

This shows each word with its part-of-speech tag.

## Noun Phrase Extraction

TextBlob can also find noun phrases using `.noun_phrases`.

```python
wiki.noun_phrases
```

Here, TextBlob identifies "python" as a noun phrase.

## Sentiment Analysis

TextBlob can analyze sentiment using polarity and subjectivity.

Polarity ranges from -1 to 1, where negative numbers are more negative and positive numbers are more positive.

Subjectivity ranges from 0 to 1, where values closer to 0 are more objective and values closer to 1 are more subjective.

I created a new TextBlob to test the sentiment features.

```python
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!")
```

```python
testimonial.sentiment
```

This gives me both the polarity and subjectivity scores.

```python
testimonial.sentiment.polarity
```

I can also just look at the polarity score by itself.

## Tokenization

TextBlob can split text into individual words or sentences.

```python
zen = TextBlob(
    "Beautiful is better than ugly. "
    "Explicit is better than implicit. "
    "Simple is better than complex."
)
```

```python
zen.words
```

This splits the text into individual words.

```python
zen.sentences
```

This splits the text into individual sentences.

```python
for sentence in zen.sentences:
    print(sentence.sentiment)
```

Since each sentence works like a TextBlob, I can also look at the sentiment of each sentence separately.

## Word Inflection and Lemmatization

TextBlob can also change words between singular and plural forms.

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
```

```python
sentence.words
```

```python
sentence.words[2].singularize()
```

This changes "spaces" into "space."

```python
sentence.words[-1].pluralize()
```

This changes "level" into "levels."

TextBlob can also lemmatize words, which basically finds a base form of the word.

```python
from textblob import Word
```

```python
w = Word("octopi")
w.lemmatize()
```

Here, "octopi" becomes "octopus."

```python
w = Word("went")
w.lemmatize("v")
```

I specified that "went" is a verb, so it returns "go."

## WordNet Integration

TextBlob can use WordNet to show different meanings of words.

```python
from textblob import Word
from textblob.wordnet import VERB
```

```python
word = Word("octopus")
word.synsets
```

This shows the different WordNet meanings for "octopus."

```python
Word("hack").get_synsets(pos=VERB)
```

Here I only looked at meanings where "hack" is being used as a verb.

```python
Word("octopus").definitions
```

I can also look at the definitions for the different meanings.

```python
from textblob.wordnet import Synset
```

```python
octopus = Synset("octopus.n.02")
shrimp = Synset("shrimp.n.03")
```

```python
octopus.path_similarity(shrimp)
```

This gives a score for how similar the two meanings are.

## WordLists

A WordList is similar to a normal Python list, but it has some extra methods for working with words.

```python
animals = TextBlob("cat dog octopus")
animals.words
animals.words.pluralize()
```

This pluralizes all of the words in the WordList.

## Spelling Correction

TextBlob can also try to correct spelling mistakes.

```python
b = TextBlob("I havv goood speling!")
print(b.correct())
```

It tries to guess what the correctly spelled words should be.

```python
from textblob import Word

w = Word("falibility")
w.spellcheck()
```

This gives a spelling suggestion and a confidence score.

## Word and Noun Phrase Frequencies

TextBlob can count how many times a word appears.

```python
monty = TextBlob(
    "We are no longer the Knights who say Ni. "
    "We are now the Knights who say Ekki ekki ekki PTANG."
)

monty.word_counts["ekki"]
```

This counts how many times "ekki" appears and is not case-sensitive.

```python
monty.words.count("ekki")
```

This is another way to count the same word.

```python
monty.words.count("ekki", case_sensitive=True)
```

This time the count is case-sensitive.

```python
wiki.noun_phrases.count("python")
```

The same type of counting can also be used with noun phrases.

## Parsing

TextBlob can also parse a sentence and give grammatical information about the words.

```python
b = TextBlob("And now for something completely different.")
print(b.parse())
```

The output gives different grammatical labels for each word.

## TextBlobs Are Like Python Strings

TextBlob objects can also work a lot like regular Python strings.

```python
zen[0:19]
```

This uses slicing just like a regular string.

```python
zen.upper()
```

This changes the whole TextBlob to uppercase.

```python
zen.find("Simple")
```

This shows the index where "Simple" starts.

```python
apple_blob = TextBlob("apples")
banana_blob = TextBlob("bananas")
apple_blob < banana_blob
apple_blob == "apples"
```

TextBlobs can be compared with other TextBlobs and normal strings.

```python
apple_blob + " and " + banana_blob
"{0} and {1}".format(apple_blob, banana_blob)
```

They can also be combined with normal strings.

## N-grams

An n-gram is a group of words that appear next to each other.

```python
blob = TextBlob("Now is better than never.")
blob.ngrams(n=3)
```

Since I used `n=3`, this makes groups of three words.

## Start and End Indices of Sentences

TextBlob can also show where each sentence starts and ends in the original text.

```python
for s in zen.sentences:
    print(s)
    print("---- Starts at index {}, Ends at index {}".format(s.start, s.end))
```

This prints each sentence and its starting and ending position in the full TextBlob.
